```python
model = ChatOpenAI(model="qwen3:8b", base_url="http://localhost:11434/v1/", api_key="")
```

## Create a LLM call function

This call function defines a system prompt to guide the LLM and takes the user prompt as an argument.

```python
def llm_prompt(user_prompt: str) -> list[str]:
    prompt = [
      {
        'role': 'system',
        'content': 'You are a Vitis HLS product expert who is very familiar with the user guide and provides the guide to the end user.',
      },
      {
        'role': 'user',
        'content': user_prompt,
      },
    ]
    return prompt
```

Let's start querying the LLM

```python
response = model.invoke(llm_prompt('How are you?'))
print(response.content)
```

Let's ask something outside its knowledge base. As you can see, the model starts hallucinating.

```python
response1 = model.invoke(llm_prompt('What Operating Systems are supported by Vitis HLS?'))
print(response1.content)
```

```python
response2 = model.invoke(llm_prompt("Where can I find tutorials and examples for Vitis HLS?"))
print(response2.content)
```

```python
response3 = model.invoke(llm_prompt('Provide the C/C++ code to write a vector add in Vitis HLS?'))
Markdown(response3.content)
```

```python
response4 = model.invoke(llm_prompt("What does the dataflow pragma do?"))
print(response4.content)
```

```python
response5 = model.invoke(llm_prompt("What interfaces does Vitis HLS support?"))
print(response5.content)
```

```python
response6 = model.invoke(llm_prompt("In what chapter can I find documentation about directives?"))
print(response6.content)
```
