# Model Serving

In this notebooks we will explore how to use an Ollama endpoint to query an LLM about topics that may not necessarily be part of its knowledge base.

```python
from ollama import chat
```

## Create a LLM call function

This call function defines a system prompt to guide the LLM and takes the user prompt as an argument.

```python
def llm_prompt(user_prompt: str) -> list[str]:
    prompt = [
      {
        'role': 'system',
        'content': 'You are a Vitis HLS product expert who is very familiar with the user guide and provides the guide to the end user.',
      },
      {
        'role': 'user',
        'content': user_prompt,
      },
    ]
    return prompt
```

Let's start querying the LLM

```python
response = chat(model='llama3.1:8b', messages=llm_prompt('How are you?'))
print(response['message']['content'])
```

Let's ask something outside its knowledge base. As you can see, the model starts hallucinating.

```python
response1 = chat(model='llama3.1:8b', messages=llm_prompt('What Operating Systems are supported by Vitis HLS?'))
print(response1['message']['content'])
```

```python
response2 = chat(model='llama3.1:8b', messages=llm_prompt("Where can I find tutorials and examples for Vitis HLS?"))
print(response2['message']['content'])
```

```python
response3 = chat(model='llama3.1:8b', messages=llm_prompt('Provide the C/C++ code to write a vector add in Vitis HLS?'))
print(response3['message']['content'])
```

```python
response4 = chat(model='llama3.1:8b', messages=llm_prompt("What does the dataflow pragma do?"))
print(response4['message']['content'])
```

```python
response5 = chat(model='llama3.1:8b', messages=llm_prompt("What interfaces does Vitis HLS support?"))
print(response5['message']['content'])
```

```python
response6 = chat(model='llama3.1:8b', messages=llm_prompt("In what chapter can I find documentation about directives?"))
print(response6['message']['content'])
```
