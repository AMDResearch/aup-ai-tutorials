## Manually defining Prompt

One of the simples way to define a prompt is a list of dictionaries where we define the prompt type using `role` and then in `content` we specify our instructions.

- `system` prompt, which is typically hidden from users and are instructions provided to the AI model before user interaction.
- `user` prompt, this is typically the questions you send to a chatbot.

In RAG systems, we can use these two prompts to our advantage.

```python
prompt_manual = [
    {
        'role': 'system',
        'content': 'You are a renowned poet',
    },
    {
        'role': 'user',
        'content': 'Why is the sky blue?',
    }
]
```

```python
prompt_manual
```

## Defining Prompt with Langchain

Lanchain provide helper functions to define these roles in a more user friendly manner. Note that `user` and `human` are the same.

```python
sysmsg = SystemMessage(content="You are a renowned poet")
humanmsg = HumanMessage(content="Why is the sky blue?")
```

```python
lc_message = [sysmsg, humanmsg]
lc_message
```

`prompt_manual` and `lc_message` are exactly the same when you invoke an LLM.

## Prompt Template

You have seen how to create simple prompts, but what if you want to update such prompt with more information, from instance from a database. You do not need to redefine the prompt, you can use a prompt template.

```python
template = ChatPromptTemplate(
    [
        ("system", "You are a renowned {job}"),
        ("human", "{question}"),
    ]
)
```

```python
prompt_value = template.invoke(
    {
        "job": "poet",
        "question": "Why is the sky blue?",
    }
)
prompt_value
```

## Prompt Template for RAG

The previous example is very simple, for RAG systems we can do a better job with the prompt.

Let's start by defining a `user_template`, where we provide instructions and left `{context}` and `{question}` as variables that will be completed when we invoke the template.

```python
user_template = """

Use the following pieces of context to answer the question at the end.

{context}

Question: {question}

Helpful Answer:
"""
```

Now, using `ChatPromptTemplate` we define the template we will pass to the LLM

```python
template = ChatPromptTemplate(
    [
        ("system", "You are a helpful assistant that answers the user questions based on the provided context. If the provided context does not contain the answer, say you don't know."),
        ("human", user_template),
    ]
)
```

Let's define some context and the question

```python
context = """
The daytime sky appears blue because air molecules scatter shorter wavelengths of sunlight more than longer ones (redder light).
The night sky appears to be a mostly dark surface or region spangled with stars.
"""
question = "Why is the sky blue?"
```

Finally, by invoking the template these variables are automatically expanded and we are ready to send this to the LLM.

```python
prompt_value = template.invoke(
    {
        "context": context,
        "question": question,
    }
)
prompt_value
```
